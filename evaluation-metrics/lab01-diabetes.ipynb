```python
zero_features = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

total_count = diabete_df['Glucose'].count()

for feature in zero_features :
	zero_count = diabete_df[diabete_df[feature] == 0][feature].count()
	print('{0} 0 건수는 {1}, 퍼센트는 {2:.2f} %'.format(feature, zero_count, 100 * zero_count/total_count))
```

```python
mean_zero_featues = diabete_df[zero_features].mean()
diabete_df[zero_features] = diabete_df[zero_features].replace(0, mean_zero_featues)
```

```python
from sklearn.linear_model import LogisticRegression

X = diabete_df.iloc[:, :-1]
y = diabete_df.iloc[:, -1]

# StandardScaler 클래스를 이용해 피처 데이터 세트에 일괄적으로 스케일링 적용
scaler = StandardScaler( ) # 표준화 적용 평균 0, 분산 1
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size = 0.2, random_state = 156, stratify=y)

# 로지스틱 회귀로 학습, 예측 및 평가 수행. 
lr_clf = LogisticRegression()
lr_clf.fit(X_train , y_train)
pred = lr_clf.predict(X_test)
```

```python
print(accuracy_score(y_test, pred))
print(recall_score(y_test, pred))
print(precision_score(y_test, pred))
print(f1_score(y_test, pred))
```
