### evaluation metric : accuracy
1. accuracy는 $\frac{correct \space predict \space count}{total \space predict \space count}$로 계산된다.
2. accuracy는 이진 분류에서 레이블의 분포가 불균형할 때 왜곡되기 쉽다. 따라서, 이진 분류기의 성능은 정확도만으로 판단하지 않는다.
3. 이진 분류의 경우 다양한 성능 지표와 함께 결합해서 사용한다.
4. API : `sklearn.metrics.accuracy_score()`

### Accuracy의 왜곡 with mnist digit dataset
* 해당 데이터를 이진 분류 데이터로 가공한 뒤에 단순히 0으로만 예측하면 정확도가 90%에 달하게 된다.

```python
from sklearn.datasets import load_digits
from sklearn.base import BaseEstimator
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import numpy as np
import pandas as pd


class MyFakeClassifier(BaseEstimator) :
	def fit(self, X, y=None) :
		pass

	def predict(self, X) :
		return np.zeros( (len(X), 1), dtype=bool)

digits = load_digits()
y = (digits.target == 7).astype(int)

X_train, X_test, y_train, y_test = train_test_split(digits.data, y, random_state=11)
```

```python
print(pd.Series(y_test).value_counts())

fakeclf = MyFakeClassifier()
fakeclf.fit(X_train, y_train)
pred = fakeclf.predict(X_test)

print(accuracy_score(y_test, pred))
```
