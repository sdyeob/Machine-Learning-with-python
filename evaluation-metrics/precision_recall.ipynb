### evaluation metric : precision과 recall
- 불균형한 레이블 클래스를 가지는 이진 분류 모델에서는 많은 데이터 중에서 중점적으로 찾아야 하는 매우 적은 수의 결과값에 Positive를 설정해 1값을 부여하고, 그렇지 않은 경우는 0 값을 부여한다.
	- 이는, Precision과 Recall의 계산식에 의해서 그런 것 같기도 하고, 그리고 관행인 것 같다.
- 두 지표 모두 높은 수치를 얻는 Estimator가 좋은 분류기로 평가된다.

* precision(정밀도) : 
	- Confusion matrix의 TP / (FP + TP)로 계산 가능하다.
	- 실제 Negative인 데이터의 예측을 Positive로 하면 안되는 경우 사용하는 지표
	- API : `sklearn.metrics.precision_score()`

* recall(재현율) : 
	- Confusion matrix의 TP / (FN + TP)로 계산 가능하다.
	- 실제 Positive인 데이터의 예측을 Negative로 하면 안되는 경우 사용하는 지표
	- API : `sklearn.metrics.recall_score()`
