### sklearn: model_selection
- sklearn의 model_selection 모듈은 다음과 같은 기능을 제공한다.
	1. train/test dataset 분리 : `sklearn.model_selection.train_test_split()`
		* 첫번째 파라미터 : 피쳐 데이터 세트
		* 두번째 파라미터 : 레이블 데이터 세트
		* 주요 옵션 : `test_size`, `shuffle(전체 데이터 섞기)`, `random_state`
	2. 교차 검증 분할 및 평가
	3. Estimator의 하이퍼 파라미터 튜닝을 위한 다양한 함수 및 클래스

### Practice 1

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

iris = load_iris()
dt_clf = DecisionTreeClassifier()

iris_data = iris.data
iris_label = iris.target

X_train, X_test, y_train, y_test = train_test_split(iris_data, iris_label, test_size=0.3, random_state=121)

dt_clf.fit(X_train, y_train)
pred = dt_clf.predict(X_test)

print(accuracy_score(y_test, pred))
```
