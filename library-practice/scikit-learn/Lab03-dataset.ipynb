### sklearn: dataset
1. 일반적으로 sklearn에서 제공하는 dataset은 dictionary형태이다.
	- 이 때, Key값은 data(feature data), target(target data), feature_names, target_names 등이 있다.
2. dataset의 key값 확인
	- `$dataset.keys()`
3. dataset의 특정 키에 대한 밸류 얻기
	- `$dataset.key`로 접근한다. ex) `$dataset.data` or `$dataset['data']`

```python
from sklearn.datasets import load_iris

iris_data = load_iris()
print(iris_data.keys()) # 해당 데이터 세트가 어떤 키값을 가지는지 확인할 수 있다.
```

```python
print("data of dataset : ", iris_data.data)
print("target of dataset : ", iris_data.target)
print("target_names of dataset : ", iris_data.target_names)
print("feature_names of dataset : ", iris_data.feature_names)
```
