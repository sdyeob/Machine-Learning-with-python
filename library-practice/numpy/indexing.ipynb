### Practice 6 : ndarray의 Indexing
1. 단일 값 인덱싱
 - 반환값은 ndarray가 아닌 단일 원소이다.
 - 단일값 인덱싱을 통해 원본 ndarray의 값을 수정할 수 있다.
 - n차원 ndarray의 단일 값 인덱싱은 []안에서 콤마로 구분하며, 각 부분은 axis를 의미한다.

2. 슬라이싱(Slicing)
 - 연속된 데이터를 한번에 인덱싱할 때 사용된다.
 - 반환값은 ndarray이다.
 - st:en를 이용해서 슬라이싱이 가능하다. 이 때, `en-1`까지 슬라이싱한다.
 - :를 기준으로 앞이나 뒤에 아무것도 적혀있지 않으면, 자동으로 start, end가 삽입된다.

3. Fancy Indexing
 - Fancy indexing은 axis n에 필요한 인덱스 집합을 리스트 혹은 ndarray로 전달해주면, 그 위치의 ndarray를 추출하는 인덱싱을 의미한다.

4. Boolean Indexing
- 비교 연산자(`<`, `<=`, `>`, `>=`, `==`, `!=`)를 이용해서 ndarray내의 조건에 맞는 값만 추출할 수 있다.
- 여러 조건을 엮는 경우에는 비트 연산자(`&`, `|`, `~`)을 이용하면 된다.
	- 단,여러 조건을 엮는 경우 비교 연산자가 비트 연산자보다 우선 순위가 높기 때문에 반드시 괄호로 우선 순위를 높혀주자.

### Practice : 단일값 Indexing

```python
import numpy as np

array1 = np.arange(start=1, stop=10) 
print(array1)

value = array1[2]
print(value)
print(type(value))

array1[0] = 0
print(array1)

array3 = array1.reshape(-1, 3)
print(array3)

array3[0, 0] = 1
print(array3)
```

### Practice : Slicing

```python
import numpy as np

array = np.arange(8)
print(array)
print(array[:-2])
```

### Practice : Fancy Indexing

```python
import numpy as np

array = np.arange(start=1, stop=10) 
array = array.reshape(3, 3)

array1 = array[[0, 1], 2]
array2 = array[[0, 1], 0:2]
array3 = array[[0, 1]]

print(f'{array1=}', end='\n\n')
print(f'{array2=}', end='\n\n')
print(f'{array3=}', end='\n\n')
```

### Practice : Boolean Indexing

```python
import numpy as np

array = np.arange(1, 10)
print(array[array > 5])

array2 = array.reshape(3, 3)
print(array2[(array2 < 4) | (array2 == 4)])
```
