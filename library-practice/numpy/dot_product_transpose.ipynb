### Practice : 행렬곱과 전치행렬 계산하기
1. `np.dot(A, B)`
	- 해당 함수를 이용하면, 행렬 A와 행렬 B의 행렬곱(내적)을 계산할 수 있다.
		- 단, 2차원인 경우 행렬 A의 열의 개수와, 행렬 B의 행의 개수가 같아야 행렬곱이 성립한다.
2. `np.transpose(ndarray)` or `ndarray.transpose()`
	- Diagonal을 기준으로 행렬을 반전시킨다.

### Practice : 행렬곱

```python
import numpy as np

A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[7, 8], [9, 10], [11, 12]])

print(np.dot(A, B))

B = B.transpose()

#print(np.dot(A, B)) # Output : ValueError: shapes (2,3) and (2,3) not aligned: 3 (dim 1) != 2 (dim 0)
```

### Practice : 행렬의 transpose

```python
import numpy as np

array = np.arange(3).reshape(-1, 1)

print(array.tolist())

t_array = np.transpose(array)

print(t_array.tolist())
print(array.tolist())
```
