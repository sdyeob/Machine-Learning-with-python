### Pandas : Column Handling
1. DataFrame에 새로운 Column 추가하기
	- `$DataFrame[new_column] = new_value`
2. 기존 Series값들을 이용해서 새로운 Series를 DataFrame에 추가하기
	- `$DataFrame[new_column] = DataFrame[A] + DataFrame[B]`
3. 칼럼 1개 이상 삭제하기
	- `$DataFrame.drop(labels=['col1', 'col2'], axis=1, inplace=True(False))`
4. 행(레코드) 1개 이상 삭제하기
	- `$DataFrame.drop(labels=[0, 1, 2], axis=0, inplace=True(False))`
* 삭제 시, inplace옵션을 조심하자. inplace=True인 경우라면 반환값은 None이다.

### Practice1 : 새로운 Series에 값 할당해서 DataFrame에 추가하기

```python
import pandas as pd

titanic_df = pd.read_csv('titanic_train.csv')

titanic_df['Age_0'] = 0
display(titanic_df.head(3))
```

### Practice2 : DataFrame내의 다른 Series들을 이용해서 새로운 Series의 값을 초기화하고, DataFrame에 추가하기

```python
titanic_df['Age_by_10'] = titanic_df['Age'] * 10
titanic_df['Family_No'] = titanic_df['SibSp'] + titanic_df['Parch'] + 1

display(titanic_df)
```

### Practice3 : 특정 Column 삭제하기

```python
drop_age_0 = titanic_df.drop(labels='Age_0', axis=1, inplace=False)
display(drop_age_0.head(3))

#drop_age_by_10 = titanic_df.drop(labels='Age_by_10', axis=1, inplace=True) # 위험한 코드 : drop_age_by_10은 None이 된다.
```

### Practice4 : 특정 row(Record) 삭제하기

```python
delete_012 = titanic_df.drop(labels=[0, 1, 2], axis=0, inplace=False)
display(delete_012.head(3))
```
