### Pandas 1: Pandas의 주요 객체 이해하기
- Pandas는 행과 열로 이루어진 2차원 데이터를 효율적으로 가공/처리할 수 있게 만드는 라이브러리이다.

1. Index
	- 개별 데이터를 고유하게 식별하는 Key값
	- Series와 DataFrame은 모두 Index를 Key값으로 가진다.
2. Series
	- 컬럼 하나를 Series라고 부른다.
	- Series의 각 행에도 고유한 Index값이 할당되어 있다.
3. DataFrame
	- 1개 이상의 Series를 가지는 데이터 구조이다.
	- 각 행을 Record라고 부르고, 각 Record에는 고유한 Index값이 할당되어 있다.

- 각 Series는 내부적으로 (Index, Data) 형태의 쌍(Pair)을 들고있다. 이 때, 두 Series를 합쳐 DataFrame을 생성하는 경우, Pandas는 두 Series의 Index 값을 비교하여 동일한 Index를 가진 데이터끼리 같은 행(Row)으로 묶어줍니다.
	- 만약 두 Series의 Index가 서로 다르다면
		1. 순서가 다를 때는 Pandas는 데이터의 위치(순서)가 아니라 Index의 값을 기준으로 병합하므로, 순서가 섞여 있어도 알아서 인덱스를 매칭하여 정렬합니다.
		2. 일부 인덱스가 누락/불일치할 때는, 두 Series가 가진 Index의 합집합(Union)을 기준으로 DataFrame의 행이 만들어집니다. 한쪽 Series에만 존재하는 인덱스가 있다면, 해당 값이 없는 쪽에는 결측치인 NaN (Not a Number)이 채워집니다.

### Pandas methods
1. `DataFrame.head(n)`
	- DataFrame의 앞에서 n개의 데이터 출력하기
	- n을 명시하지 않는 경우, default로 n = 5이다.

2. `DataFrame.shape`
	- DataFrame의 행과 열의 개수를 튜플 형태로 반환

3. `DataFrame.info()`
	- 해당 DataFrame의 구조를 보여준다.
	- RangeIndex : 전체 몇개의 entries(record, row)가 있고, 몇개의 Feature(column)을 가지는지 보여주는 항목
	- 각 컬럼의 이름, 특정 컬럼의 non-null인 데이터의 개수, 특정 컬럼이 저장하는 데이터의 타입 등의 정보를 보여준다.

4. `DataFrame.describe()`
	- 숫자형 컬림의 통계적인 특성을 조사하는 메서드로, object 타입의 컬럼은 출력에서 제외된다.
	- 출력하는 정보는 n-percentile 분포도, 평균값, 최댓값, 최솟값을 보여준다.
	- 숫자형 카테고리 컬럼 여부를 판단하는 경우 유용하다.

5. `DataFrame.value_counts()` or `Series.value_counts()`
	- 해당 시리즈에 저장된 값의 종류와 각 값의 개수를 출력해준다.
	- 이 때, 값의 개수가 큰 종류부터 내림차순으로 정렬된 Series를 반환한다.
	- 기본적으로 해당 메서드는 Nan값은 drop해서 카운팅한다. 주의하도록 하자.
