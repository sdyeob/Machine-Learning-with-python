### sklearn : 데이터 전처리
- 기본적으로 결손값(NaN, Null)은 허용되지 않는다.
	1. 피쳐 값 중 결손값이 적다면 피쳐의 평균으로 간단하게 대체할 수 있다.
	2. 피쳐 값 중 결손값이 많다면, 드롭하는게 더 낫다.
	3. 피쳐에 결손값이 많지만, 중요한 피쳐라면 어떤 값으로 결손값을 대체할지 잘 결정해야한다.

- sklearn의 머신러닝 알고리즘은 문자열 값을 입력으로 받지 않는다. 따라서, 문자열 값을 인코딩을 통해 숫자형 값으로 변경해주어야 한다.
	1. label encoding
		- ML알고리즘에 숫자의 크고 작음에 대한 특성이 작용하는 경우, label encoding이용 시 예측 성능이 떨어질 수 있다.
		- 선형 회귀에는 사용 X -> 숫자 인코딩이 영향을 끼칠 수 있음
		- 트리 계열에는 사용 O -> 숫자의 이런 특성을 반영 X
	2. one-hot encoding
		- label encoding의 숫자의 크기 관련 문제를 해결하기 위해 나온 encoding 방식
		- 결과가 희소 행렬로 나오므로, 밀집 행렬로 변환이 필요함
	3. feature vectorization

### Practice 1: LabelEncoder
- fit(), transform()을 통해 encoding 수행
- 숫자에 대응되는 값은 객체의 classes_인스턴스 변수를 이용하면 알 수 있다.
- inverse_transform()을 통해 decoding 가능

```python
from sklearn.preprocessing import LabelEncoder

items = ['TV', '냉장고', '전자레인지', '컴퓨터', '선풍기', '선풍기', '믹서', '믹서']

le = LabelEncoder()
le.fit(items)
labels = le.transform(items)

print(labels)
print(le.classes_)
print(decoded_labels := le.inverse_transform(labels))
```

### practice : One Hot Encoder
- fit()메서드의 인자로 반드시 2차원 데이터를 제공해야한다.
- transform()메서드의 결과는 2차원 행렬에서 '1'이 어떤 위치에 있는지에 대한 정보다. 이를, 2차원 행렬 형태로 변환하기 위해서는 출력값에 toarray()메서드를 이용해서 변환해야 한다.

```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

items = ['TV', '냉장고', '전자레인지', '컴퓨터', '선풍기', '선풍기', '믹서', '믹서']

oh_encoder = OneHotEncoder()
print(items)
print(np.array(items).reshape(-1, 1))

dim2_items = np.array(items).reshape(-1, 1)
oh_encoder.fit(dim2_items)

oh_items = oh_encoder.transform(dim2_items)

print(oh_items)
print(oh_items.toarray())
```

### practice : pandas를 이용한 One Hot Encoding
- pd.get_dummies(df)를 이용하면, 간단하게 one-hot encoding이 가능하다.

```python
import pandas as pd

df = pd.DataFrame({'items':['TV', '냉장고', '전자레인지', '컴퓨터', '선풍기', '선풍기', '믹서', '믹서'] })

print(pd.get_dummies(df).astype('Int8'))
```
