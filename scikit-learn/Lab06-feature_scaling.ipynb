### sklearn : feature scaling
- feature scaling : 서로 다른 변수의 값 범위를 일정한 수준으로 맞추는 작업으로. 표준화와 정규화 방식이 있다.
- Scaler 클래스의 fit, transform은 2차원 이상의 데이터에 대해서만 가능하다.
1. 표준화(Standardization) : 데이터의 피쳐 각각이 평균이 0이고, 분산이 1인 가우시안 정규 분포를 가진 값으로 변환하는 작업
	- sklearn에서 구현한 RBF커널을 이용하는 SVM(Support Vector Machine), 선형회귀, 로지스틱 회귀의 경우 데이터가 가우시안 분포를 가지고 있다고 가정하기 때문에, 데이터를 표준화하는게 중요하다.
	- Question : Data의 분포가 정규 분포인 경우에만 사용할 수 있는건가? -> 맞음, 데이터가 정규 분포가 아니라면 단순히 스케일링만 될 뿐, 표준 정규 분포의 형태로 분포가 바뀌는게 아님.

2. 정규화(Normalization) : 서로 다른 피쳐의 크기(다루는 숫자의 크기)를 통일하기 위해 크기를 변환하는 개념
	- 모두 동일한 크기 단위로 비교하기 위해서 [0, 1]사이의 값으로 변환하는 것.
		- 여기서 크기 단위란 각 피쳐에서의 값의 크기(1, 2 등의 숫자)가 다뤄지는 의미가 다르다.

3. fit, transform, fit_transform
	- `Scaler.fit(data)` : Scaler가 이용하는 공식의 상수값을 정하기 위해서 전체 데이터를 한번 훑는 과정
		적용 공식 $$z = \frac{x - \mu}{\sigma} $$ $$x_{scaled} = \frac{x - x_{min}}{x_{max} - x_{min}} $$ 이 두 공식들을 적용하기 위해서 `fit(data)` 시 평균 ($\mu$)• 분산 및 표준편차 ($\sigma$)• 최솟값 ($x_{min}$,)• 최댓값 ($x_{max}$) 을 계산해둔다.

4. 주의점
	- `Scaler.fit()`을 train에 적용하면, 상수가 train 데이터셋에 맞게 설정된다. 이후 test에도 해당 메서드를 적용하면, 상수가 다시 test 데이터셋에 맞게 설정된다. 그러면, 두 데이터셋의 스케일링 단위가 바뀐다. 따라서, `Scaler.fit()`은 보통 train 데이터셋에 단 한번만 적용한다.
	- 데이터를 train, test로 split하기 전에, fit -> transform을 마친 뒤에 split하자.

### Practice : StandardScaler

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

iris = load_iris()
iris_data = iris.data
iris_features = iris.feature_names
iris_df = pd.DataFrame(data=iris_data, columns=iris_features)

print('= ' * 20)
print('Before SS')
df = pd.DataFrame(data={'mean':iris_df.mean().tolist(), \
						'var':iris_df.var().tolist()}, \
				index=iris.feature_names)
print(df)

ss = StandardScaler()
ss.fit(iris_data)
standardization_data = ss.transform(iris_data)

print('= ' * 20)
print('After SS')
df2 = pd.DataFrame(data={'mean':standardization_data.mean().tolist(), \
						 'var':standardization_data.var().tolist()}, \
				index=iris.feature_names)
print(df2)

sns.scatterplot(iris_data)
```

### practice : MinMaxScaler

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

iris = load_iris()
iris_data = iris.data

mms = MinMaxScaler()
mms.fit(iris_data)
iris_scaled = mms.transform(iris_data)

iris_df_scaled = pd.DataFrame(data=iris_scaled, columns=iris.feature_names)
print(iris_df_scaled.min())
print(iris_df_scaled.max())
```
