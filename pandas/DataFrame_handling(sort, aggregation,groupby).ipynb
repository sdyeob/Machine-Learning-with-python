### Pandas : Sort, Aggregation, GroupBy
1. `$DataFrame.sort_values()` : 
	- parameters : by=[columns], inplace=False, ascending=False0

2. Aggregation(데이터 집계) : 여러 개의 레코드 값을 요약하여 하나의 통계값(합계, 평균, 최댓값 등)으로 변환하는 작업
	- `$DataFrame.count()`, `$DataFrame.sum()`, `$DataFrame.max()` 등
	- 특정 컬럼으로 필터링을 한 뒤에 사용하는게 아니라면, DataFrame내의 모든 컬럼에 대해서 적용한다.
	

3. `$DataFrame.GroupBy()`
	- 해당 메서드 호출 시 Index가 Group기준이 된다.
	- 반환값은 Grouping된 DataFrame인 `DataFrameGroupBy`인스턴스이다.
	- `DataFrameGroupBy` 인스턴스에 인덱싱을 활용해서 원하는 컬럼만 뽑아낸 뒤 Aggregation을 수행하면 좋다.

### Practice : DataFrame을 특정 열을 기준으로(by) 정렬하기

```python
import pandas as pd

titanic_df = pd.read_csv('titanic_train.csv')
sorted_df = titanic_df.sort_values(by=['Age'], inplace=False, ascending=True)

display(sorted_df.head(3))

sorted_df2 = titanic_df.sort_values(by=['Pclass', 'Age'], inplace=False, ascending=True)
display(sorted_df2.head())
```

### Practice : Data Aggregation 수행하기
- count(), max(), min() 등

```python
import pandas as pd

titanic_df = pd.read_csv('titanic_train.csv')

print(titanic_df.count(), end='\n\n')

print(titanic_df[['Age', 'Fare']].max())
```

### Practice: GroupBy() 적용

```python
import pandas as pd

titanic_df = pd.read_csv('titanic_train.csv')
titanic_df.groupby(by='Pclass').count()

titanic_df.groupby(by='Pclass')[['Age', 'Fare']].count()

titanic_df.groupby(by='Pclass')['Age'].agg([max, min])


agg_format = {'Age':'max', 'SibSp': 'sum', 'Fare':'mean'}
titanic_df.groupby(by='Pclass').agg(agg_format)
```
