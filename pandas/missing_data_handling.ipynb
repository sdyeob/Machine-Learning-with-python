### pandas: missing data handling
1. DataFrame내에 결측치가 있는지 확인
	- `$DataFrame.isna().sum()`
2. DataFrame내에서 결측치가 있는 행 혹은 열을 제거
	- `$DataFrame.dropna()`
3. DataFrame내에서 결측치를 특정 값으로 채우기
	- `$DataFrame.fillna(value)`

### practice : DataFrame내 결측치 확인

```python
import pandas as pd

titanic_df = pd.read_csv('titanic_train.csv')
titanic_df.info() # method 1
print(titanic_df.isna().sum()) # method 2
```

### practice : DataFrame내에서 결측치가 있는 행 혹은 열을 제거


```python
titanic_df2 = titanic_df.dropna()
titanic_df2.info()
```

### practice : DataFrame내에서 결측치를 특정 값으로 채우기

```python
titanic_df['Age'].fillna(titanic_df['Age'].mean())
```
